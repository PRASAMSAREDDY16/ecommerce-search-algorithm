```python
products = [
    {
        "name": "iPhone 15",
        "price": 79999,
        "category": "Electronics",
        "rating": 4.8
    },
    {
        "name": "Samsung Galaxy",
        "price": 64999,
        "category": "Electronics",
        "rating": 4.5
    },
    {
        "name": "Nike Shoes",
        "price": 4999,
        "category": "Fashion",
        "rating": 4.3
    },
    {
        "name": "Dell Laptop",
        "price": 55000,
        "category": "Electronics",
        "rating": 4.6
    },
    {
        "name": "Adidas T-Shirt",
        "price": 1999,
        "category": "Fashion",
        "rating": 4.1
    }
]


def search_products(products, category, max_price, min_rating):

    results = []

    for product in products:

        # Category filter
        if product["category"] != category:
            continue

        # Price filter
        if product["price"] > max_price:
            continue

        # Rating filter
        if product["rating"] < min_rating:
            continue

        # Product matches all conditions
        results.append(product)

    return results


results = search_products(
    products,
    "Electronics",
    70000,
    4.5
)


for product in results:
    print(product["name"])
```
